#  Step 1: Download Research Papers using pygetpapers

In this step, we use **pygetpapers**, a command-line tool that retrieves open-access research articles from Europe PMC.

### Command Structure

pygetpapers -q "search query" -k number_of_papers -o output_folder -x --pdf

### Explanation of Parameters

- **-q** → Search query (keywords connected using OR / AND)
- **-k** → Number of papers to download
- **-o** → Output directory where results will be saved
- **-x** → Download full-text XML files
- **--pdf** → Download PDF versions (if available)

This step will:
- Search Europe PMC database
- Download metadata
- Retrieve XML full-text articles
- Retrieve PDFs (if open access)

The downloaded files will be stored inside the specified output folder.
