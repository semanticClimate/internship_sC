```python
import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud
from collections import Counter

def generate_wordcloud(csv_path, label):
    df = pd.read_csv(csv_path)
    counter = Counter()

    # Access the first column by its integer index, which is more robust
    # than relying on a specific column name like '0'.
    for row in df.iloc[:, 0].dropna():
        items = set(x.strip() for x in str(row).split(",") if x.strip())
        counter.update(items)

    if not counter: # Check if the counter is empty
        print(f"No words found for {label}. Skipping word cloud generation.")
        return

    wc = WordCloud(
        width=900,
        height=450,
        background_color="white",
        colormap="viridis"
    ).generate_from_frequencies(counter)

    plt.figure(figsize=(12,6))
    plt.imshow(wc)
    plt.axis("off")
    plt.show()

    out = csv_path.replace(".csv", "_wordcloud.png")
    wc.to_file(out)
    print(f"Saved: {out}")

for d in CONFIG["DICTIONARIES"]:
    path = f"/content/{PROJECT}/{d.lower()}.csv"
    generate_wordcloud(path, d)
```

```python
import requests, time

wikidata_cache = {}

def fetch_wikidata(text):
    if pd.isna(text):
        return None

    ids = []
    for name in set(x.strip() for x in str(text).split(",") if x.strip()):
        if name in wikidata_cache:
            ids.append(wikidata_cache[name])
            continue

        r = requests.get(
            "https://www.wikidata.org/w/api.php",
            params={
                "action": "wbsearchentities",
                "search": name,
                "language": "en",
                "format": "json",
                "limit": 1
            },
            timeout=10
        )
        qid = r.json()["search"][0]["id"] if r.json().get("search") else None
        wikidata_cache[name] = qid
        if qid:
            ids.append(qid)
        time.sleep(0.3)

    return ", ".join(ids) if ids else None
```

```python
for d in CONFIG["DICTIONARIES"]:
    path = f"/content/{PROJECT}/{d.lower()}.csv"
    df = pd.read_csv(path)
    # Access the first column by its integer index to avoid KeyError
    df["Wikidata_ID"] = df.iloc[:, 0].apply(fetch_wikidata)
    df.to_csv(path.replace(".csv", "_with_wikidata.csv"), index=False)
```

```python
import networkx as nx
import re, os

G = nx.Graph()

for d in CONFIG["DICTIONARIES"]:
    df = pd.read_csv(f"/content/{PROJECT}/{d.lower()}_with_wikidata.csv")

    for _, row in df.iterrows():
        src = re.search(r"(PMC\\d+)", str(row["file_path"]))
        source = src.group(1) if src else "DOC"

        for target in str(row["0"]).split(","):
            t = target.strip()
            if t:
                G.add_edge(source, t, dict=d)

nx.write_graphml(G, f"/content/{PROJECT}/network.graphml")
print("✔ Network saved for Cytoscape")
```
