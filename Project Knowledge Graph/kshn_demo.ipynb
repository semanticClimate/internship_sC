## Step 2: Data Import
Upload the source HTML files to the Colab environment for processing.

```python
from google.colab import files
uploaded = files.upload()
input_filename = list(uploaded.keys())[0]
print("done")
```

## Step 3: Graph Generation
Execute the `kshn` pipeline to extract entities and build relationships using Wikidata and local co-occurrence analysis.

```python
use_all_features = True #@param {type:"boolean"}
export_graphml = True #@param {type:"boolean"}
custom_output_name = "" #@param {type:"string"}

cmd = f"python kshn.py -i {input_filename}"
if use_all_features: cmd += " --all"
if export_graphml: cmd += " --graphml"
if custom_output_name: cmd += f" -o {custom_output_name}"

!{cmd}
print("done")
```

## Step 4: Export & Download
Download the resulting Knowledge Graph (HTML) and optional data files (GraphML) for local use.

```python
import os
from google.colab import files

if custom_output_name:
    out_file = custom_output_name
else:
    out_file = input_filename.rsplit('.', 1)[0] + "_KG.html"

if os.path.exists(out_file):
    files.download(out_file)
    graphml_file = out_file.rsplit('.', 1)[0] + ".graphml"
    if os.path.exists(graphml_file): 
        files.download(graphml_file)
print("done")
```
